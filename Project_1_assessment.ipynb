## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Function: Validate read length and character composition.
    code stops checking when it is not validated"""
    upper_seq = sequence.upper()
    errors = []
    if len(upper_seq) != len(reference):
        errors.append("wrong_length")

    for char in upper_seq:
        if char not in "ACGT":
            errors.append("invalid_characters")
            break

    return {
        "sequence": upper_seq,
        "valid": len(errors) == 0,
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """This is a function that finds mismatches between the sequence and reference.

    Arguments:
    sequence: The read sequence
    reference: The reference sequence

    Returns:
    A list of each mismatch is printed or [] if they match.
    """
    upper_seq = sequence.upper()
    upper_ref = reference.upper()
    mutations = []
    for i, (ref_base, read_base) in enumerate(zip(upper_ref, upper_seq), start=1):
        if ref_base != read_base:
            mutations.append({
                "position": i,
                "reference_base": ref_base,
                "read_base": read_base
            })
    return find_mutations(read_seq, reference_seq)

    assert mutations == [{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]

    print(mutations)
````

### 🎨 Design your solution

You must decide:

### 💻 Enter your code here

````python
def summarise_samples(analysed_reads):
    """Summarise mismatch rates, counts, substitutions, and top positions per sample."""
    if not analysed_reads:
        return {}
    summary = {}
    for read in analysed_reads:
        name, group = read["sample"], read["group"]

        if name not in summary:
            summary[name] = {
                "group": group, "valid_reads": 0, "invalid_reads": 0,
                "mismatch_reads": 0, "mismatch_percentage": 0.0, "mismatch_events": 0,
                "substitution_counts": {}, "position_counts": {}, "top_positions": []
            }

        s = summary[name]

        if read["status"] == "valid":
            s["valid_reads"] += 1
            muts = read["mutations"]

            if len(muts) > 0:
                s["mismatch_reads"] += 1
            s["mismatch_events"] += len(muts)

            for m in muts:
                sub = m["reference_base"] + ">" + m["read_base"]
                s["substitution_counts"][sub] = s["substitution_counts"].get(sub, 0) + 1

                pos = m["position"]
                s["position_counts"][pos] = s["position_counts"].get(pos, 0) + 1
        else:
            s["invalid_reads"] += 1

    for name, s in summary.items():
        v = s["valid_reads"]
        s["mismatch_percentage"] = (s["mismatch_reads"] / v) * 100 if v > 0 else None

        # Sorting for top positions
        pos_list = list(s["position_counts"].keys())
        pos_list.sort(key=lambda p: (-s["position_counts"][p], p))
        s["top_positions"] = pos_list[:3]

    return summary
````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

Each list must contain **one percentage per biological sample**.

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

````python
# Write your code here.

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    if not sample_summary:
        return []

    combined_position_counts = {}
    for sample_name, data in sample_summary.items():
        group = data["group"]
        if group == control_group or group == treated_group:
            for pos, count in data["position_counts"].items():
                if pos not in combined_position_counts:
                    combined_position_counts[pos] = 0
                combined_position_counts[pos] += count

    all_positions = list(combined_position_counts.keys())
    all_positions.sort(key=lambda p: (-combined_position_counts[p], p))
    top_positions = all_positions[:3]

    results = []

    for pos in top_positions:
        control_percentages = []
        treated_percentages = []
        excluded_samples = []

        for sample_name, data in sample_summary.items():
            group = data["group"]
            if group != control_group and group != treated_group:
                continue
            v_reads = data["valid_reads"]
            if v_reads == 0:
                excluded_samples.append(sample_name)
                continue
            mismatches = data["position_counts"].get(pos, 0)
            percentage = (mismatches / v_reads) * 100

            if group == control_group:
                control_percentages.append(percentage)
            elif group == treated_group:
                treated_percentages.append(percentage)

        print(f"---Position {pos}---")
        if excluded_samples:
            print(f"Excluded samples due to 0 valid reads: {', '.join(excluded_samples)}")

        print(f" {control_group} percentages: {[round(p, 2) for p in control_percentages]}")
        print(f" {treated_group} percentages: {[round(p, 2) for p in treated_percentages]}")

        stat_value = None
        p_val = None

        if "calculate_statistics" in globals():
            stat_result = calculate_statistics(control_percentages, treated_percentages)
            if stat_result is not None:
                stat_value = stat_result.get("statistic")
                p_val = stat_result.get("p_value")
                print(f" Welch T-test -> Statistic: {stat_value:.4f}, p-value: {p_val:.4f}")
            else:
                print(" Welch T-test -> Could not be calculated, no data.")
        else:
            print(" (calculate_statistics function not found)")
        print("-" * 40)

        pos_result = {
            "position": pos,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": (sum(control_percentages) / len(control_percentages)) if control_percentages else None,
            "treated_mean": (sum(treated_percentages) / len(treated_percentages)) if treated_percentages else None,
            "difference": None,
            "statistic": stat_value,
            "p_value": p_val
        }
        if pos_result["control_mean"] is not None and pos_result["treated_mean"] is not None:
            pos_result["difference"] = pos_result["treated_mean"] - pos_result["control_mean"]

        results.append(pos_result)

    return results
````

````python
# Enter your code here
import csv

def validate_read(sequence, reference):
    """Normalise the sequence to uppercase and check for length or character errors."""
    upper_seq = sequence.upper()
    errors = []
    if len(upper_seq) != len(reference):
        errors.append("wrong_length")
    for char in upper_seq:
        if char not in "ACGT":
            errors.append("invalid_characters")
            break
    return {
        "sequence": upper_seq,
        "valid": len(errors) == 0,
        "errors": errors
    }

def find_mutations(sequence, reference):
    """Find mismatch positions (1-indexed) between sequence and reference."""
    upper_seq = sequence.upper()
    upper_ref = reference.upper()
    mutations = []
    for i, (ref_base, read_base) in enumerate(zip(upper_ref, upper_seq), start=1):
        if ref_base != read_base:
            mutations.append({
                "position": i,
                "reference_base": ref_base,
                "read_base": read_base
            })
    return mutations

def analyse_reads(reads, reference):
    """Analyse all reads using validate_read and find_mutations."""
    results = []
    for read in reads:
        validation = validate_read(read["sequence"], reference)
        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(validation["sequence"], reference)
        else:
            status = "invalid"
            mutations = []
        analysed_read = {
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": validation["errors"],
            "mutations": mutations
        }
        results.append(analysed_read)
    return results

def summarise_samples(analysed_reads):
    """Summarise mismatch rates, counts, substitutions, and top positions."""
    if not analysed_reads:
        return {}
    summary = {}
    for read in analysed_reads:
        name, group = read["sample"], read["group"]

        if name not in summary:
            summary[name] = {
                "group": group, "valid_reads": 0, "invalid_reads": 0,
                "mismatch_reads": 0, "mismatch_percentage": 0.0, "mismatch_events": 0,
                "substitution_counts": {}, "position_counts": {}, "top_positions": []
            }

        s = summary[name]

        if read["status"] == "valid":
            s["valid_reads"] += 1
            muts = read["mutations"]

            if len(muts) > 0:
                s["mismatch_reads"] += 1
            s["mismatch_events"] += len(muts)

            for m in muts:
                sub = m["reference_base"] + ">" + m["read_base"]
                s["substitution_counts"][sub] = s["substitution_counts"].get(sub, 0) + 1

                pos = m["position"]
                s["position_counts"][pos] = s["position_counts"].get(pos, 0) + 1
        else:
            s["invalid_reads"] += 1

    for name, s in summary.items():
        v = s["valid_reads"]
        s["mismatch_percentage"] = (s["mismatch_reads"] / v) * 100 if v > 0 else None

        # Simple sorting for top positions
        pos_list = list(s["position_counts"].keys())
        pos_list.sort(key=lambda p: (-s["position_counts"][p], p))
        s["top_positions"] = pos_list[:3]

    return summary

# Data loader to prevent NameError
try:
    _reference = reference
    _reads = reads
except NameError:
    print("Loading dataset files.")
    with open("reference.txt") as f:
        lines = [ln.strip() for ln in f if ln.strip()]
        if lines[0].startswith(">"):
            lines = lines[1:]
        _reference = "".join(lines).upper()

    with open("reads.csv", newline="") as f:
        _reads = list(csv.DictReader(f))

#Execution Code
print("Processing biological sequencing datasets")
analysed_test_reads = analyse_reads(_reads, _reference)
summary_result = summarise_samples(analysed_test_reads)

#Step 4 Summary Table
print("\n" + "="*50)
print("STEP 4: SAMPLE METRICS SUMMARY")
print("="*50)
for sample_name, info in summary_result.items():
    print(f"Sample: {sample_name} ({info["group"].upper()})")
    print(f"  - Valid Reads: {info["valid_reads"]}")
    print(f"  - Invalid Reads: {info["invalid_reads"]}")
    print(f"  - Total Mismatch Events: {info["mismatch_events"]}")
    print(f"  - Mismatch Percentage: {info["mismatch_percentage"]:.2f}%" if info["mismatch_percentage"] is not None else "  - Mismatch Percentage: N/A")
    print(f"  - Substitution Counts: {info["substitution_counts"]}")
    print(f"  - Top Mismatched Positions: {info["top_positions"]}")
    print("-" * 30)

#Step 5 Comparative Analysis
print("\n" + "="*50)
print("STEP 5: POSITION EXPLORATION & STATISTICAL COMPARISONS")
print("="*50)
compare_positions(summary_result)
````
